```python
#we may need some code in the ../python directory and/or matplotlib styles
import sys
sys.path.append('../python/')

#matplotlib for plotting
import matplotlib as mpl
from matplotlib import pyplot as plt
plt.style.use('mplstyles/stylelib/standard.mplstyle')

#other computational libraries
import numpy as np
```

```python
#define x1 and x2
x1=0
x2=1
y1=0
y2=1
```

```python
def l(y,a,b):
    return a*y+b

def q(y,a,b):
    return a*y**2 + b

def t(y,a,b):
    return a*y**3 + b
```

```python
A1 = np.array([[y1,1],[y2,1]])
A2 = np.array([[y1**2,1],[y2**2,1]])
A3 = np.array([[y1**3,1],[y2**3,1]])

B = np.array([x1,x2])

C1 = np.linalg.solve(A1,B)
C2 = np.linalg.solve(A2,B)
C3 = np.linalg.solve(A3,B)

print(C1)
print(C2)
print(C3)
```

```python
import scipy as sp

def dydx(y,x,c):
    return np.sqrt(x/(c-x))

xv = np.linspace(x1,x2,100)
sol = sp.integrate.odeint(dydx,y1,xv,args=(1.5,))
print(sol[:,0])
```

```python
#set up a 1d plot
fig,axes = plt.subplots(1,1,figsize=(9.0,8.0),sharex=True)
ax1 = axes

yvec=np.linspace(y1,y2,100)

ax1.plot(yvec,-l(yvec,C1[0],C1[1]),color='g',linestyle="-",linewidth=2, \
         label='linear')
ax1.plot(yvec,-q(yvec,C2[0],C2[1]),color='m',linestyle="-",linewidth=2, \
         label='quadratic')




ax1.set_yscale('linear')
#ax1.set_yscale('log')
ax1.set_xlim(y1,y2) 
ax1.set_ylim(-x2*1.5,x1)
ax1.set_xlabel(r'y [m]')
ax1.set_ylabel('-x  [m]')
ax1.grid(True)
ax1.yaxis.grid(True,which='minor',linestyle='--')
ax1.legend(loc=1,prop={'size':12})
#ax1.legend(bbox_to_anchor=(1.04,1),borderaxespad=0,prop={'size':22})

for axis in ['top','bottom','left','right']:
  ax1.spines[axis].set_linewidth(2)

plt.tight_layout()
#plt.savefig('figures/figure.png')
plt.show()
```
